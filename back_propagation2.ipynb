#### 练习题：双参数模型训练

##### 1. 模型定义

模型：

$ \hat{y}=w_2(w_1x+1) $

其中：

- x：输入数据。
- $w_1$,$w_2$：两个需要训练的模型参数。
- $\hat y$：预测结果。

初始条件：

$ x=3,\quad y=11 $

$ w_1=1,\quad w_2=2 $

##### 2. 损失函数

采用带有二分之一系数的平方误差损失：

$ L=\frac{1}{2}(\hat{y}-y)^2 $

这里的 $y$ 是真实标签，$\hat y$ 是模型预测值。

##### 3. 学习率

$ \eta=0.01 $

```python
import torch

x = torch.tensor(3.0)
y = torch.tensor(11.0)

w1 = torch.tensor(1.0, requires_grad=True)
w2 = torch.tensor(2.0, requires_grad=True)

eta = 0.01

y_pred = (x * w1 + 1) * w2
print(f"y_pred = (x * w1 + 1) * w2 = {y_pred.item()}")

loss = (y_pred - y) ** 2 * 0.5
print(f"loss = (y_pred - y) ** 2 * 0.5 = {loss.item()}")
```

先定义中间变量：

$$
a=w_1x+1
$$

$$
\hat y=aw_2
$$

损失函数：

$$
L=\frac{1}{2}(\hat y-y)^2
$$

对 w1 求偏导：

$$
\frac{\partial L}{\partial w_1}
=
\frac{\partial L}{\partial \hat y}
\times
\frac{\partial \hat y}{\partial a}
\times
\frac{\partial a}{\partial w_1}
=
(\hat y-y)w_2x
= 
-18
$$

对 w2 求偏导：

$$
\frac{\partial L}{\partial w_2}
=
\frac{\partial L}{\partial \hat y}
\times
\frac{\partial \hat y}{\partial w_2}
=
(\hat y-y)a
=
-12
$$

然后梯度下降就行了

```python
loss.backward()

assert w1.grad is not None
assert w2.grad is not None

print(f"w1偏导：{w1.grad.item()}")
print(f"w2偏导：{w2.grad.item()}")

with torch.no_grad():
    w1 -= eta * w1.grad 
    w2 -= eta * w2.grad
```
