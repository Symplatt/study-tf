#### 1. 模型定义

模型：

$$
\hat{y}=wx
$$

其中：

- $x$：输入数据，例如房屋面积。
- $w$：模型参数，需要训练。
- $\hat{y}$：模型预测结果。
- $y$：真实房价。

假设：

$$
x=3,\quad y=7,\quad w=2
$$

#### 2. 前向传播

$$
\hat{y}=wx=2\times3=6
$$

#### 3. 计算损失

定义平方误差损失：

$$
L=(\hat{y}-y)^2
$$

代入：

$$
L=(6-7)^2=1
$$

#### 4. 反向传播

计算损失对参数 $w$ 的梯度：

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat{y}}
\cdot
\frac{\partial \hat{y}}{\partial w}
$$

代入：

$$
=2(2\times3-7)\times3=-6
$$

#### 5. 梯度下降

假设学习率为：

$$
\eta=0.1
$$

参数更新公式：

$$
w_{\text{new}}=w_{\text{old}}-\eta\frac{\partial L}{\partial w}
$$

代入：

$$
w_{\text{new}}=2-0.1\times(-6)=2.6
$$

```python
# TODO：画出计算图

import torch

eta = 0.1

x = torch.tensor(3.0)
y = torch.tensor(7.0)

w = torch.tensor(2.0, requires_grad=True)

y_pred = w * x

loss = (y_pred - y) ** 2
print(f"loss = {loss}")

loss.backward()

assert w.grad is not None
print(f"w.grad = {w.grad.item()}")

w_old = w.item()
with torch.no_grad():
    w -= eta * w.grad
    print(f"下一个step的w = {w_old} - {w_old} * {eta} = {w.item()}")
```
